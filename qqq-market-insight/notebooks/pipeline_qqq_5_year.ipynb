## 8 Model training and held-out evaluation

`pipeline.build_models()` in the reference script constructs and fits four fixed models on the training period only. The table below exposes the fitted estimator classes and main settings, then shows held-out metrics. The output label `GradientBoosting` specifically refers to `HistGradientBoostingClassifier`, not the older `GradientBoostingClassifier`. No hyperparameter search is performed.

```python
x_train = train_data[pipeline.FEATURES]
y_train = train_data["target_up"]
x_test = test_data[pipeline.FEATURES]
y_test = test_data["target_up"]

fitted_models, model_metrics = pipeline.build_models(x_train, y_train, x_test, y_test, config=config)
model_description = {
    'Majority baseline': 'Always predicts the most frequent training class',
    'Logistic regression': 'Linear classifier; median imputation and scaling are fitted inside its Pipeline',
    'Random forest': 'Constrained ensemble of decision trees',
    'GradientBoosting': 'Histogram-based gradient-boosted trees; report focus',
}
model_settings = []
for model_name, fitted_model in fitted_models.items():
    parameters = fitted_model.get_params()
    if model_name == 'Logistic regression':
        settings = f"C={parameters['model__C']}, class_weight={parameters['model__class_weight']}, max_iter={parameters['model__max_iter']}"
    elif model_name == 'Random forest':
        settings = f"n_estimators={parameters['n_estimators']}, max_depth={parameters['max_depth']}, min_samples_leaf={parameters['min_samples_leaf']}"
    elif model_name == 'GradientBoosting':
        settings = f"learning_rate={parameters['learning_rate']}, max_iter={parameters['max_iter']}, max_depth={parameters['max_depth']}, min_samples_leaf={parameters['min_samples_leaf']}"
    else:
        settings = f"strategy={parameters['strategy']}"
    estimator_class = ('Pipeline → ' + type(fitted_model.named_steps['model']).__name__) if model_name == 'Logistic regression' else type(fitted_model).__name__
    model_settings.append({'Output label': model_name, 'Estimator class': estimator_class, 'Purpose': model_description[model_name], 'Key settings': settings})
display(pd.DataFrame(model_settings))
assert type(fitted_models['GradientBoosting']).__name__ == 'HistGradientBoostingClassifier'

metric_columns = ["accuracy", "balanced_accuracy", "precision", "recall", "f1", "roc_auc"]
display(
    model_metrics.style
    .format({column: "{:.4f}" for column in metric_columns})
    .highlight_max(subset=metric_columns, color="#D9EAD3")
)

# Reconcile this notebook run against the saved CLI table when it exists.
cli_metrics_path = pipeline.TABLES_PATH / 'model_metrics.csv'
if cli_metrics_path.exists():
    cli_metrics = pd.read_csv(cli_metrics_path).set_index('model').loc[model_metrics['model']]
    notebook_metrics = model_metrics.set_index('model').loc[cli_metrics.index]
    metric_differences = (notebook_metrics[metric_columns] - cli_metrics[metric_columns]).abs()
    maximum_difference = metric_differences.to_numpy().max()
    assert np.allclose(notebook_metrics[metric_columns], cli_metrics[metric_columns], rtol=0, atol=1e-12), 'Notebook metrics differ from saved CLI metrics; check raw data, config, code, and library versions.'
    print(f'Notebook and saved CLI metrics agree across all four models (maximum absolute difference {maximum_difference:.2g}).')
else:
    print(f'No saved CLI metrics at {cli_metrics_path}; run python src/pipeline_qqq_5_year.py to compare.')
```

## 9 Expanding time-series cross-validation

```python
time_series_cv = TimeSeriesSplit(n_splits=pipeline.CV_SPLITS, gap=pipeline.CV_GAP)
cv_result = cross_validate(
    clone(main_model),
    x_train,
    y_train,
    cv=time_series_cv,
    scoring=["accuracy", "balanced_accuracy", "f1", "roc_auc"],
    n_jobs=-1,
    error_score="raise",
)

cv_folds = pd.DataFrame(
    {key.removeprefix("test_"): value for key, value in cv_result.items() if key.startswith("test_")}
)
cv_folds.insert(0, "fold", range(1, len(cv_folds) + 1))

boundaries = []
for train_indices, validation_indices in time_series_cv.split(x_train):
    boundaries.append(
        {
            "train_rows": len(train_indices),
            "validation_rows": len(validation_indices),
            "train_end_date": train_data.iloc[train_indices[-1]]["Date"].date().isoformat(),
            "validation_start_date": train_data.iloc[validation_indices[0]]["Date"].date().isoformat(),
            "gap_rows": int(validation_indices[0] - train_indices[-1] - 1),
        }
    )
cv_folds = pd.concat([cv_folds, pd.DataFrame(boundaries)], axis=1)

cv_summary = pd.DataFrame(
    [
        {"metric": metric, "mean": cv_folds[metric].mean(), "standard_deviation": cv_folds[metric].std(ddof=1)}
        for metric in ("accuracy", "balanced_accuracy", "f1", "roc_auc")
    ]
)
display(cv_folds.style.format({metric: "{:.4f}" for metric in ("accuracy", "balanced_accuracy", "f1", "roc_auc")}))
display(cv_summary.style.format({"mean": "{:.4f}", "standard_deviation": "{:.4f}"}))
```

## 10 Held-out permutation feature importance

```python
permutation = permutation_importance(
    main_model,
    x_test,
    y_test,
    n_repeats=config["permutation_repeats"],
    random_state=config["random_state"],
    scoring="balanced_accuracy",
    n_jobs=-1,
)

importance = pd.DataFrame(
    {
        "feature": pipeline.FEATURES,
        "importance_mean": permutation.importances_mean,
        "importance_std": permutation.importances_std,
    }
).sort_values("importance_mean", ascending=False)
family_lookup = {
    feature: family
    for family, features in pipeline.FEATURE_GROUPS.items()
    for feature in features
}
importance["family"] = importance["feature"].map(family_lookup)

display(importance.head(12).style.format({"importance_mean": "{:.4f}", "importance_std": "{:.4f}"}))

top_importance = importance.head(12).sort_values("importance_mean")
fig, axis = plt.subplots(figsize=(9, 6))
axis.barh(
    top_importance["feature"],
    top_importance["importance_mean"],
    xerr=top_importance["importance_std"],
    color="#2F80ED",
    alpha=0.85,
)
axis.axvline(0, color="black", linewidth=0.8)
axis.set_title(f"Held-out permutation importance for {target_model_name}")
axis.set_xlabel("Decrease in balanced accuracy after permutation")
axis.set_ylabel("Feature")
plt.tight_layout()
plt.show()

display(Markdown("Error bars show ±1 standard deviation across permutations; they are not confidence intervals. Correlated indicators can divide importance between related features."))
```

### Feature-family aggregation

```python
family_importance = (
    importance.groupby("family", as_index=False)
    .agg(importance_mean=("importance_mean", "sum"), feature_count=("feature", "count"))
    .sort_values("importance_mean", ascending=False)
)
display(family_importance.style.format({"importance_mean": "{:.4f}"}))

ordered_family = family_importance.sort_values("importance_mean")
fig, axis = plt.subplots(figsize=(8, 4.8))
axis.barh(ordered_family["family"], ordered_family["importance_mean"], color="#219EBC")
axis.axvline(0, color="black", linewidth=0.8)
axis.set_title("Indicator-family permutation importance")
axis.set_xlabel("Sum of individual held-out importance means")
axis.set_ylabel("Feature family")
plt.tight_layout()
plt.show()

display(Markdown("Family values are sums of individual-feature importance means and are not joint family permutation tests."))
```
